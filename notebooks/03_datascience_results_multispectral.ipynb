# multispectral_results

`GET /datascience_results/multispectral_results/`

Multispectral forest-health raster outputs indexed by site and period, with the S3 and HTTPS location of each cloud-optimized raster.

Written by `veritree_multispectral_foresthealth :: task_publish_geospatial_service` into `multispectral.tbl_raster_results`.

**Needs:** `AWS_RDS_*` in `.env` pointing at the analytics database.

> **Answered by a live S3 scan.** Pass `live_scan=true` — this route indexes the rasters
> straight out of S3, so a result appears as soon as the pipeline writes it. Without
> the flag the route returns **501**: the database source is not implemented.
> The response reports its origin in `source_type`.


```python
import json
import os
import sys

import requests
from dotenv import load_dotenv

# Run against a locally started service:  python3 src/main.py pipeline pipeline_api
sys.path.insert(0, os.path.abspath(".."))
load_dotenv(os.path.join("..", ".env"))

BASE_URL = os.getenv("VT_API_BASE_URL", "http://localhost:8000")
TOKEN = os.getenv("API_ENDPOINT_TOKEN")
HEADERS = {"Token": TOKEN}

assert TOKEN, "API_ENDPOINT_TOKEN missing -- copy env_template to .env and fill it in"
print(f"target: {BASE_URL}")
```

```python
# Confirm the service is up before sending anything to the route.
try:
    health = requests.get(f"{BASE_URL}/health", timeout=5)
    print(health.status_code, health.json())
except requests.exceptions.ConnectionError:
    print("Service is not running. Start it with:\n"
          "    python3 src/main.py pipeline pipeline_api")
```

## Request

All filters are optional; omit them to page through everything. `limit` and `offset` are applied by the database, so the response is bounded by the page rather than the table.

> **Answered by a live S3 scan.** Pass `live_scan=true` — this route indexes the rasters
> straight out of S3, so a result appears as soon as the pipeline writes it. Without
> the flag the route returns **501**: the database source is not implemented.
> The response reports its origin in `source_type`.


```python
params = {
    "live_scan": "true",   # REQUIRED: the database source is not implemented
    "country": None,
    "site": None,
    "subsite": None,
    "period": None,
    "raster": None,        # e.g. PRODUCTIVITY, STRUCTURE, WATER_STATUS, CONNECTIVITY
    "limit": 100,
    "offset": 0,
}

response = requests.get(f"{BASE_URL}/datascience_results/multispectral_results/",
                        params={k: v for k, v in params.items() if v is not None},
                        headers=HEADERS, timeout=60)
```

```python
print(response.status_code)
if response.ok:
    body = response.json()
    print(json.dumps(body, indent=2)[:2000])
else:
    print(response.text[:1000])
```

## Paging

`n_records` is the full count from a `COUNT(*)`; `n_returned` is the size of this page. Walk the result set by advancing `offset`.

> **Answered by a live S3 scan.** Pass `live_scan=true` — this route indexes the rasters
> straight out of S3, so a result appears as soon as the pipeline writes it. Without
> the flag the route returns **501**: the database source is not implemented.
> The response reports its origin in `source_type`.


```python
all_rows, offset, page_size = [], 0, 500

while True:
    r = requests.get(f"{BASE_URL}/datascience_results/multispectral_results/",
                     params={"limit": page_size, "offset": offset},
                     headers=HEADERS, timeout=60)
    if not r.ok:
        print(r.status_code, r.text[:300]); break
    body = r.json()
    all_rows.extend(body["results"])
    offset += page_size
    if offset >= body["n_records"] or body["n_returned"] == 0:
        break

print(f"fetched {len(all_rows)} rows")
```
